```python
df['Дата открытия, категориальный'] = df['Дата открытия, категориальный'].astype('category')
df['Торговая площадь, категориальный'] = df['Торговая площадь, категориальный'].astype('category')
df['Регион'] = df['Регион'].astype('category')
df['Населенный пункт'] = df['Населенный пункт'].astype('category')
df.dtypes
```

```python
pto = df.groupby('Месяц', observed=True)['РТО'].mean()

plt.figure(figsize=(8, 5))
plt.plot(pto.index, pto.values)
plt.xlabel("Месяц")
plt.ylabel("Средний РТО")
plt.title("Средний РТО по месяцам")
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(df["РТО"], bins=50)
plt.xlabel("РТО")
plt.title("Распределение РТО")
plt.show()
```

```python
df["rto_lag_1"] = df.groupby("new_id")["РТО"].shift(1)
df["rto_lag_2"] = df.groupby("new_id")["РТО"].shift(2)
df["rto_lag_3"] = df.groupby("new_id")["РТО"].shift(3)

df["rto_mean_3"] = (
    df.groupby("new_id")["РТО"]
    .shift(1)
    .rolling(window=3)
    .mean()
)
```

```python
feature_cols = df.drop(columns=['РТО']).columns
X = df[feature_cols]
y_log = np.log1p(df['РТО'])

november = df[df['Месяц'] == 10].copy()
november['Месяц'] = 11
X_november = november[feature_cols]
```

## XGBoost

```python
import xgboost as xgb
```

```python
xgb_model = xgb.XGBRegressor(
    n_estimators=1000,
    learning_rate=0.11,
    max_depth=10,
    random_state=1,
    enable_categorical=True,
    objective='reg:squarederror',
)

xgb_model.fit(X, y_log, verbose=False)

xgb_pred = np.expm1(xgb_model.predict(X_november))
```

```python
xgb_submission = pd.DataFrame({
    'new_id': november['new_id'].values,
    'rto': xgb_pred,
})

xgb_submission.to_csv('test.csv', index=None)
```

## Catboost (хуже)

```python
import catboost as cat
```

```python
df.dtypes
```

```python
cat_model = cat.CatBoostRegressor(
    n_estimators=1000,
    learning_rate=0.11,
    max_depth=10,
    random_state=1,
)

cat_model.fit(X, y_log, verbose=False, cat_features=[2, 3, 4, 5])

cat_pred = np.expm1(cat_model.predict(X_november))
```

```python
cat_submission = pd.DataFrame({
    'new_id': november['new_id'].values,
    'rto': cat_pred,
})

cat_submission.to_csv('test_cat.csv', index=None)
```
